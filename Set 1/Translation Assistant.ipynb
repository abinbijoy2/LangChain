```python
from langchain_databricks import ChatDatabricks
from langchain_core.prompts import PromptTemplate

# Initialize llm
llm = ChatDatabricks(
    endpoint="databricks-meta-llama-3-3-70b-instruct",
    temperature=0.1,
    max_tokens=1000
)

# Define translation prompt
translation_prompt = PromptTemplate(
    input_variables=["text", "target_language"],
    template="""You are a professional translator.

Your task: Translate the following text into {target_language}.

Important instructions:
- Provide ONLY the translation.
- Maintain the tone and style of the original text.
- Do not add explanations, intro, or concluding remarks.
- Preserve all original formatting and punctuation.

Text to translate:
{text}

Translation:"""
)

print("✓ Translation components initialized successfully!")
```

```python
# Example: English to Spanish
text = "Hello, how are you today? I hope you're having a great day!"
target_lang = "Spanish"

prompt = translation_prompt.format(text=text, target_language=target_lang)
result = llm.invoke(prompt)

import textwrap
print(f"\nExample: English → {target_lang}")
print("=" * 30)
print(f"Original: {text}")
wrapped_translation = textwrap.fill(result.content)
print(f"Translation: {wrapped_translation}\n")
```

```python
dbutils.widgets.text("input_text", " ", "Enter text to translate:")

dbutils.widgets.dropdown(
    "target_lang", "German",
    ["Spanish", "French", "German", "Italian", "Portuguese", "Japanese", "Chinese", "Korean", "Russian", "Arabic", "Hindi", "Malayalam"],
    "Select target language:"
)

print("✓ Widgets created!")
```

```python
# Get values from widgets
user_text = dbutils.widgets.get("input_text")
selected_language = dbutils.widgets.get("target_lang")

if user_text.strip() and user_text != "Enter text to translate:":
    print("\n🔄 Translating text...\n")
    
    prompt = translation_prompt.format(text=user_text, target_language=selected_language)
    result = llm.invoke(prompt)
    
    print("=" * 30)
    print("INPUT DETAILS")
    print("=" * 30)
    print(f"Original Text: {user_text}")
    print(f"Target Language: {selected_language}")
    
    print("\n" + "=" * 30)
    print("✨ TRANSLATION RESULT")
    print("=" * 30)
    wrapped_result = textwrap.fill(result.content)
    print(f"\n{wrapped_result}\n")
```
