```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory
from langchain_databricks import ChatDatabricks

llm = ChatDatabricks(
    endpoint="databricks-meta-llama-3-3-70b-instruct",
    temperature=0.7,
    max_tokens=500
)

# Create conversation memory
memory = ConversationBufferMemory()

# Create conversation chain
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False
)
```

```python
dbutils.widgets.text("user_message", "Hello! How are you?", "💬 Your Message:")

user_input = dbutils.widgets.get("user_message")

import textwrap
if user_input.strip():
    print(f"👤 You: {user_input}")
    print("-"*90)
    
    bot_response = conversation.predict(input=user_input)
    
    print(textwrap.fill(f"🤖 Bot: {bot_response}", width=100))
    print("\n" + "="*90)

else:
    print("⚠️ Please enter a message in the widget above.")
```

```python
print("\n CONVERSATION HISTORY STORED IN MEMORY:")
print("=" * 60)
print(memory.buffer)
print("=" * 60)
```

```python
# Clear the conversation memory
memory.clear()

print("✅ Memory cleared successfully!")
print("🚀 You can now start a fresh conversation!")
```
