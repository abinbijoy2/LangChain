```python
from langchain_databricks import ChatDatabricks
from langchain_core.prompts import PromptTemplate

# Initialize the LLM
llm = ChatDatabricks(
    endpoint="databricks-meta-llama-3-3-70b-instruct",
    temperature=0.2,
    max_tokens=500
)

prompt_template = """
You are a code generation assistant. Generate clean, working code.

Problem: {problem_description}
Language: {language}

Output simple working codes with minimal inline comments. No explanations or descriptions needed.
"""

code_prompt = PromptTemplate(
    input_variables=["problem_description", "language"],
    template=prompt_template
)

print("✓ Prompt template created")
```

```python
#Example
problem_description = "Write a simple hello world function"
language = "Python"

formatted_prompt = code_prompt.format(
    problem_description=problem_description,
    language=language
)

result = llm.invoke(formatted_prompt)
print(result.content)
```

```python
dbutils.widgets.text("problem_input", "", "Enter Your Problem Description")
dbutils.widgets.dropdown("language_input", "Python", ["Python", "JavaScript", "Java", "C++"], "Programming Language")

user_problem = dbutils.widgets.get("problem_input")
user_language = dbutils.widgets.get("language_input")

if user_problem or user_problem.strip() != "":
    print(f"Problem Description:  {user_problem}")
    print("\n" + "="*30 + "\n")
    print(f"Generating {user_language} Code...\n")
    
    formatted_prompt = code_prompt.format(
        problem_description=user_problem,
        language=user_language
    )
    result = llm.invoke(formatted_prompt)
    print(result.content)
    
```
