```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory
from langchain_databricks import ChatDatabricks
print("✓ Libraries imported successfully!")
```

```python
# Initialize the LLM using Databricks Foundation Model API
# Using Meta Llama 3.3 70B - a powerful open-source model hosted on Databricks
llm = ChatDatabricks(
    endpoint="databricks-meta-llama-3-3-70b-instruct",
    temperature=0.7,
    max_tokens=500
)

# Create conversation memory
# ConversationBufferMemory stores the entire conversation history
memory = ConversationBufferMemory()

# Create the conversation chain
conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False  # Set to False to hide the conversation chain details
)

print("✓ Chatbot with memory created successfully!")
print("\nMemory type: ConversationBufferMemory")
print("This will remember all previous exchanges in the conversation.")
```

```python
# View the stored conversation history
print("\n📜 CONVERSATION HISTORY STORED IN MEMORY:")
print("=" * 60)
print(memory.buffer)
print("=" * 60)

# You can also access the messages
print("\n📦 Memory Variables:")
print(memory.load_memory_variables({}))
```

```python
# 🔄 CLEAR MEMORY AND START A NEW CONVERSATION
# Run this cell when you want to reset the chatbot's memory

# Clear all widgets
dbutils.widgets.removeAll()

# Clear the conversation memory
memory.clear()

print("✅ Memory cleared successfully!")
print("\n🎆 The chatbot has forgotten all previous conversations.")
print("🚀 You can now start a fresh conversation!")
print("\n💡 TIP: Run the 'Real-Time Chat with Widget' cell to start chatting again.")
```
